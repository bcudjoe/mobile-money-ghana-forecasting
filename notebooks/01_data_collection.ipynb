# Loading the raw data

This notebook pulls in the five source files and checks their shapes before any cleaning. The downloads themselves are done by hand from the Bank of Ghana portal, World Bank Findex, the NCA and the World Bank population and account-ownership series (links are in the README), so here I just read what is already sitting in `data/raw`.

## Setup (local or Google Colab)

On Colab this pulls the project data and code from GitHub. Locally it uses the repo folder you already have.

```python
# Works whether you run this locally or on Google Colab.
import os, sys
IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB and not os.path.isdir('data'):
    import subprocess
    subprocess.run(['git','clone','-q','https://github.com/bcudjoe/mobile-money-ghana-forecasting.git'])
    os.chdir('mobile-money-ghana-forecasting')
print('Colab' if IN_COLAB else 'local')
```

```python
import pandas as pd
RAW = 'data/raw/'
files = ['monthly_series.xlsx','findex_ghana_2021.xlsx','findex_ghana_2025.xlsx',
         'agent_density_population.xlsx','yearly_account_ownership.xlsx']
for f in files:
    df = pd.read_excel(RAW + f)
    print(f'{f:34} {df.shape[0]:>5} rows x {df.shape[1]:>2} cols')
```
